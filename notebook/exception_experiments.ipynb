```python
logger = CustomLogger().get_logger("ExceptionExperiment")
```

```python
class DocumentPortalException(Exception):
    """Custom exception for Document Portal errors."""
    def __init__(self, error_message:Exception,error_details:sys):# type: ignore
        _,_,exc_tb = error_details.exc_info()
        self.file_name = exc_tb.tb_frame.f_code.co_filename # type: ignore
        self.line_number = exc_tb.tb_lineno # type: ignore
        self.error_message = str(error_message)
        self.traceback_str = ''.join(traceback.format_exception(*error_details.exc_info()))
    def __str__(self):
        return f"""
        Error occurred in file [{self.file_name} at line {self.line_number}]
        Message : {self.error_message}
        Traceback:\n{self.traceback_str}
        """
    

if __name__ == "__main__":
    try:
        a = 1 / 0  # This will raise a ZeroDivisionError
        print(a)
    except Exception as e:
        # Capture the exception details
      app_exc = DocumentPortalException(e, sys)
      logger.error(app_exc)
      raise app_exc
```

```python
import sys
import traceback

try:
    1 / 0
except Exception as e:
    exc_info = sys.exc_info()  # returns (type, value, traceback)
    # Unpack exc_info and pass to format_exception
    tb_str = ''.join(traceback.format_exception(*exc_info))
    print(tb_str)
```
