### Logging Levels in Python

Python logging levels (from lowest to highest severity):

* DEBUG: Detailed information, typically for diagnosing problems.
* INFO: Confirmation that things are working as expected.
* WARNING: Something unexpected happened, or indicative of some problem.
* ERROR: A more serious problem; the program may not be able to perform some function.
* CRITICAL: A very serious error; the program itself may be unable to continue running.

You set the level using logging.basicConfig(level=logging.INFO) or similar.

If you set level to INFO, you will see INFO, Warning, error and critical messages but not DEBUG messages.
We can see messages with more severity than the level we set. But not the ones with lower severity.

```python
# Configure logging
logging.basicConfig(
    filename=LOG_PATH,
    level=logging.INFO,
    format="[%(asctime)s] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s",
)
```

### Structure the log in a specific format using structlog

```python
import structlog

log = structlog.get_logger()
```

```python
log.info('User uploaded a file to the server',
             user_id=12345,
             file_name='report.pdf',
             file_size=2048,)
```

```python
log.error('File failed to upload',
          user_id=12345,
          file_name='report.pdf',
          error='File size exceeds limit',
          file_size=4096,
          max_size=2048)
```
